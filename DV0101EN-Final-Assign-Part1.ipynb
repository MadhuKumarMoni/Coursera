# Data Description

The dataset used for this visualization assignment contains *automobile_sales* data representing automobile sales and related variables during recession and non-recession period. 

The dataset includes the following variables:
<br>1. Date: The month end date of sales observation.
<br>2. Recession: A binary variable indicating recession perion; 1 means it was recession, 0 means it was normal.
<br>3. Automobile_Sales: The number of vehicles sold during the period.
<br>4. GDP: The per capita GDP value in USD.
<br>5. Unemployment_Rate: The monthly unemployment rate.
<br>6. Consumer_Confidence: A synthetic index representing consumer confidence, which can impact consumer spending and automobile purchases.
<br>7. Seasonality_Weight: The weight representing the seasonality effect on automobile sales during the period. This variable represents the seasonal effect on automobile sales for a given month. In the automobile industry, sales often fluctuate throughout the year due to seasonal patterns—for example, sales may increase during the festive season or year-end promotions, and decrease during off-peak months like post-holiday winter periods.
A value greater than 1 indicates higher-than-average sales expected for that month due to seasonal trends (e.g., holiday season or new model launches).
A value less than 1 suggests lower-than-average sales due to seasonal slowdowns.
A value around 1 means the season has neutral or average effect on sales.
<br>8. Price: The average vehicle price during the period.
<br>9. Advertising_Expenditure: The advertising expenditure of the company.
<br>10.Vehicle_Type: The type of vehicles sold; Supperminicar, Smallfamiliycar,                 Mediumfamilycar, Executivecar, Sports.
<br>11.Competition: The measure of competition in the market, such as the number of competitors or market share of major manufacturers.
<br>12.Month: Month of the observation extracted from Date..
<br>13.Year: Year of the observation extracted from Date.
<br>
By examining various factors mentioned above from the dataset, you aim to gain insights into how recessions impacted automobile sales for your company.

####  TASK 1.2: How do trends in advertising expenditure correlate with automobile sales during non-recession periods, and what insights can be derived from this relationship.


<details><summary>Click here for a hint</summary>

<p>
    You will require:-
<br>Start by filtering the dataset to include only non-recession periods where the "Recession" column has a value of 0.
<br>Group the data by Year and calculate the average of "Automobile_Sales" and "Advertising_Expenditure".
<br>Ensure you reset the index after aggregation so the DataFrame remains structured properly.
<br>Use sns.lineplot() to plot trends:
<br>One line for average automobile sales (solid green line with circular markers).
<br>Another line for average advertising expenditure (dashed blue line with square markers).
<br>Add labels and a title to make the visualization clear.
<br>Include a legend and a grid for better readability.

````python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Filter dataset for non‑recession years
non_recession_df = df[df['Recession'] == 0]

# 2. Group by year and compute average automobile sales and advertising expenditure
trend_df = (
    non_recession_df
    .groupby('Year')[['Automobile_Sales', 'Advertising_Expenditure']]
    .mean()
    .reset_index()
)

# 3. Plot using seaborn
plt.figure(figsize=(12, 6))

# Line for automobile sales
sns.lineplot(
    data=trend_df,
    x='Year',
    y='Automobile_Sales',
    marker='o',
    color='green',
    label='Average Automobile Sales',
    linewidth=2
)

# Line for advertising expenditure
sns.lineplot(
    data=trend_df,
    x='Year',
    y='Advertising_Expenditure',
    marker='s',
    color='blue',
    linestyle='--',
    label='Average Advertising Expenditure',
    linewidth=2
)

# Labels and title
plt.title('Advertising Expenditure vs Automobile Sales (Non‑Recession Periods)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Average Values', fontsize=12)

# Improve readability
plt.grid(True, which='major', linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()

plt.show()
````

````python

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# --- 1. Split the dataset into recession and non‑recession ---
recession_df = df[df['Recession'] == 1]
non_recession_df = df[df['Recession'] == 0]

# --- 2. Compute average annual sales per vehicle type ---
recession_trend = (
    recession_df
    .groupby(['Year', 'Vehicle_Type'])['Automobile_Sales']
    .mean()
    .reset_index()
)

non_recession_trend = (
    non_recession_df
    .groupby(['Year', 'Vehicle_Type'])['Automobile_Sales']
    .mean()
    .reset_index()
)

# --- 3. Create the plot ---
plt.figure(figsize=(14, 6))

# Left subplot: Recession period
plt.subplot(1, 2, 1)
sns.lineplot(
    data=recession_trend,
    x='Year',
    y='Automobile_Sales',
    hue='Vehicle_Type',
    marker='o'
)
plt.title("Sales Trend per Vehicle Type (Recession Period)")
````

### Now you want to compare the sales of different vehicle types during a recession and a non-recession period
<br>We recommend that you use the functionality of **Seaborn Library** to create this visualization


<details><summary>Click here for a hint</summary>

<p>
     To visualize sales of different vehicles during recession and non-recession periods, you can use a bar chart
        <br> You will need to group Recession, Vehicle_Type for average Automobile_Sales and then plot it<br>
    Make use of sns.barplot(x=x,y=y, data = df)</p>
</details>


````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# --- 1) Prepare data ---------------------------------------------------------
# Keep only the columns we need
cols = ['Vehicle_Type', 'Recession', 'Automobile_Sales', 'Year']
missing = set(cols) - set(df.columns)
if missing:
    raise ValueError(f"Missing required columns in df: {missing}")

# Aggregate: average sales per vehicle type and period (recession vs non‑recession)
agg = (
    df.groupby(['Vehicle_Type', 'Recession'])['Automobile_Sales']
      .mean()
      .reset_index()
      .rename(columns={'Automobile_Sales': 'Avg_Sales'})
)

# Optional: sort vehicle types by non‑recession sales (so ordering is meaningful)
order = (
    agg[agg['Recession'] == 0]
    .sort_values('Avg_Sales', ascending=False)['Vehicle_Type']
    .tolist()
)
# Fallback ordering if there were no non-recession rows for some reason
if not order:
    order = sorted(agg['Vehicle_Type'].unique())

# Map recession flag to human‑readable labels for legend
agg['Period'] = agg['Recession'].map({0: 'Non‑Recession', 1: 'Recession'})

# --- 2) Plot (Grouped Bar Chart) --------------------------------------------
sns.set(style='whitegrid')
plt.figure(figsize=(12, 6))

ax = sns.barplot(
    data=agg,
    x='Vehicle_Type',
    y='Avg_Sales',
    hue='Period',
    order=order,
    palette={'Non‑Recession': '#1f77b4', 'Recession': '#2ca02c'},
    edgecolor='black'
)

# Add value labels on top of bars
for p in ax.patches:
    height = p.get_height()
    if pd.notna(height):
        ax.annotate(f'{height:,.0f}',
                    (p.get_x() + p.get_width()/2., height),
                    ha='center', va='bottom', fontsize=9, xytext=(0, 3),
                    textcoords='offset points')

# Aesthetics
ax.set_title('Average Automobile Sales by Vehicle Type:\nRecession vs Non‑Recession', fontsize=14)
ax.set_xlabel('Vehicle Type', fontsize=12)
ax.set_ylabel('Average Automobile Sales', fontsize=12)
ax.legend(title='Period')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
````

### TASK 1.4: Use sub plotting to compare the variations in GDP during recession and non-recession period by developing line plots for each period.
<br>Now, you want to find more insights from the data to understand the reason. <br>Plot a two line charts using subplotting to answer:-
#### How did the GDP vary over time during recession and non-recession periods? 
<br>Make use of <u>add_subplot()</u> from Matplotlib for this comparision.


````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Split the dataset into recession and non‑recession groups
recession_df = df[df['Recession'] == 1]
non_recession_df = df[df['Recession'] == 0]

# 2. Aggregate GDP values yearly
recession_gdp = (
    recession_df.groupby('Year')['GDP']
    .mean()
    .reset_index()
)

non_recession_gdp = (
    non_recession_df.groupby('Year')['GDP']
    .mean()
    .reset_index()
)

# 3. Create subplot figure
fig = plt.figure(figsize=(14, 6))

# --- Subplot 1: GDP trend during recession ---
ax1 = fig.add_subplot(1, 2, 1)
sns.lineplot(
    data=recession_gdp,
    x='Year',
    y='GDP',
    marker='o',
    color='red',
    ax=ax1
)
ax1.set_title("GDP Trend During Recession")
ax1.set_xlabel("Year")
ax1.set_ylabel("Average GDP")
ax1.grid(True, linestyle="--", alpha=0.5)

# --- Subplot 2: GDP trend during non‑recession ---
ax2 = fig.add_subplot(1, 2, 2)
sns.lineplot(
    data=non_recession_gdp,
    x='Year',
    y='GDP',
    marker='o',
    color='green',
    ax=ax2
)
ax2.set_title("GDP Trend During Non‑Recession")
ax2.set_xlabel("Year")
ax2.set_ylabel("Average GDP")
ax2.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
````

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Filter only non‑recession periods
non_recession_df = df[df['Recession'] == 0]

# 2. Aggregate average sales and seasonality weight per month
bubble_df = (
    non_recession_df
    .groupby('Month')[['Automobile_Sales', 'Seasonality_Weight']]
    .mean()
    .reset_index()
)

# Optional: order months properly if dataset uses abbreviations
month_order = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
bubble_df['Month'] = pd.Categorical(bubble_df['Month'], categories=month_order, ordered=True)
bubble_df = bubble_df.sort_values('Month')

# 3. Create Bubble Plot
plt.figure(figsize=(12, 6))

sns.scatterplot(
    data=bubble_df,
    x='Month',
    y='Automobile_Sales',
    size='Seasonality_Weight',
    sizes=(100, 2000),     # bubble size range
    alpha=0.6,
    color='teal',
    edgecolor='black'
)

# 4. Titles and labels
plt.title('Seasonality impact on Automobile Sales', fontsize=14)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Average Automobile Sales', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
````
